## 1. Ranked Actions + Reason Codes
We map model scores and empirical search signals to actionable content playbooks and clear diagnostic reason codes:

- **`DECAY_POSITION_SLIP`**: Significant rank loss over 30 days vs historical baseline ($\Delta \ge 1.2$).
- **`CTR_UNDERPERFORMING`**: Observed CTR is $25\%+$ below baseline power-law curve expectation ($\text{CTR}_{\text{exp}} = 0.3382 \cdot p^{-1.2429}$).
- **`HIGH_IMP_LOW_CLICK`**: Large impression share ($>2000$) with poor click conversion.
- **`MONITOR_VOLATILITY`**: High position rank volatility ($\sigma > 2.0$).
- **`STABLE_PERFORMER`**: Strong rankings, low volatility, optimal CTR.

### Archetype to Action Mapping Matrix:
| Content Archetype | Primary Action Playbook | Urgency |
| :--- | :--- | :--- |
| **Decaying (1)** | `REWRITE_CONTENT_INTENT` / `UPDATE_FRESHNESS` | **CRITICAL** / **HIGH** |
| **CTR Underperformer (2)** | `OPTIMIZE_METADATA_TITLES` | **HIGH** |
| **Growing / Momentum (3)** | `MONITOR_QUERY_INTENT` | **MEDIUM** |

## 2. Intended Use and Limits
### Intended Use:
- Operational decision support for SEO content strategy teams to prioritize high-ROI editorial refreshes, title tag optimizations, and content updates.
- Enables data-driven resource allocation based on expected traffic recoverability.

### Limitations & Honest Framing:
- **Observed vs Causal**: The model identifies historical performance decay, NOT causal search engine algorithm rules.
- **No Causal Refresh Guarantee**: Refreshing content does not guarantee instant rank recovery.
- **Public Compliance**: Zero raw URLs, client names, credentials, or private query text.

## 3. Human Review + The NO-GO List
### Human-in-the-Loop Editorial Protocol:
All automated recommendations MUST be reviewed by a human subject-matter expert before content modification.

### ❌ The NO-GO List (What Should NOT Be Automated):
1. **No Automated AI Content Rewriting or Direct Publishing**: Never auto-generate and auto-publish content directly without human editing.
2. **No Algorithmic Bulk 301 Redirects or Page Deletions**: Prevents accidental traffic destruction.
3. **No Unreviewed YMYL (Medical / Legal / Financial) Edits**: High-stakes regulatory compliance requires human expertise.
4. **No Brand Positioning / Core Messaging Overhauls**: Maintains corporate brand identity.

## 4. Monitoring / Retrain Triggers & Cost-Value Thinking
### Cost / Value Prioritization Framework:
$$\text{Value Score} = \text{Impressions} \times \text{CTR Deficit Ratio} \times \text{Opportunity Score}$$
Resources are focused first on pages with high impression share and severe CTR gaps to maximize return on editorial effort.

### Monitoring & Retrain Triggers:
- **Data Drift Trigger**: Rolling mean position drift shifts by $>15\%$.
- **Concept Drift Trigger**: Model ROC-AUC on rolling 30-day window drops below $0.85$.
- **Search Engine Core Update Trigger**: Retrain model after major search core updates or quarterly seasonal shifts.
