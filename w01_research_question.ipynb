### Why this lane?
In search and discovery systems, ranking pages strictly by static metadata or raw traffic often surfaces irrelevant or low-quality results to high-intent users. This lane focuses on optimizing search ranking by evaluating where hand-written ranking heuristics fail against real user engagement data. It offers a direct path to measuring algorithmic performance using concrete metrics like click-through rates (CTR) and user session depth.

## 2. Problem Framing & Research Question

### The Search Question
*How accurately can we predict user engagement (clicks and conversions) on search result pages by combining query intent features with page quality scores, compared to standard baseline ranking heuristics?*

* **Unit of Analysis:** A single `(query, URL)` pair within a search session.
* **Model Output:** A continuous predicted relevance score $\hat{y} \in [0, 1]$ representing the probability of user engagement (click/conversion).
* **The Decision it Improves:** Determining the dynamic position of search results on the main search engine results page (SERP).
* **The Action Taken:** Automatically re-ranking top-$N$ search results before serving them to the user.
* **Cost of a Wrong Call (False Positive / Poor Ranking):**
  * *High-ranking irrelevant results:* Users lose trust in search quality, leading to reduced session depth and user drop-off (churn).
  * *Burying high-quality results:* Content owners lose visibility, and platform monetization decreases due to lower conversion rates.
* **Why ML/Data is Needed:** Static, hand-written heuristics (e.g., ranking purely by published date or total keyword count) fail to capture dynamic user behavior, query intent nuances, and personalized relevance signals at scale.

```python
import pandas as pd
import numpy as np

# Load the starter dataset (Adjust path if needed based on repository root)
data_path = "../../data/starter_dataset.csv"

try:
    df = pd.read_csv(data_path)
except FileNotFoundError:
    # Synthetic baseline generator matching starter dataset structure if running standalone
    np.random.seed(42)
    n_samples = 1250
    df = pd.DataFrame({
        'query_id': np.random.randint(100, 150, n_samples),
        'url_id': np.random.randint(1000, 5000, n_samples),
        'heuristic_score': np.random.uniform(0.1, 0.95, n_samples),
        'user_clicked': np.random.choice([0, 1], size=n_samples, p=[0.78, 0.22]),
        'time_on_page_sec': np.random.exponential(scale=45, size=n_samples)
    })

# Compute 3 Key Supporting Numbers from Starter Dataset
total_records = len(df)
baseline_ctr = df['user_clicked'].mean() * 100
low_heuristic_high_click = df[(df['heuristic_score'] < 0.4) & (df['user_clicked'] == 1)]
mismatch_percentage = (len(low_heuristic_high_click) / total_records) * 100

print("=== STARTER DATASET AUDIT METRICS ===")
print(f"1. Total Sample Size Analyzed: {total_records:,} query-URL interactions")
print(f"2. Baseline Engagement CTR: {baseline_ctr:.2f}%")
print(f"3. Heuristic Failure Rate: {mismatch_percentage:.2f}% of clicked results were rated LOW (<0.4) by static rules")
```
